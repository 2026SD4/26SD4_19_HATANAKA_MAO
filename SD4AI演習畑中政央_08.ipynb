```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

prompt = ChatPromptTemplate.from_messages([
    ("system", "ユーザーが入力した料理のレシピを考えてください。"),
    ("human", "{dish}"),
])
print(prompt.to_json())
```

```python
model = ChatGoogleGenerativeAI(
    model = "gemini-flash-latest",
    google_api_key = GOOGLE_API_KEY,
    temperature = 0.7
)
```

```python
chain = prompt | model
ai_message = chain.invoke( {"dish":"焼きそば"})
print( ai_message.content)
```

```python
ai_message = chain.invoke({"dish":"りんごあめ"})
print(ai_message.content)
```

```python
from langchain_core.tools import structured
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel, Field
from typing import List

class Recipe(BaseModel):
  ingredients: List[str] = Field(description="料理の材料リスト")
  steps: List[str] = Field(description="調理手順のリスト")

prompt = ChatPromptTemplate.from_messages([
    ("system","ユーザーが入力した料理のレシピを指定された形式で考えてください。"),
    ("human","{dish}")
])
```

```python
model_gemini = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    google_api_key=GOOGLE_API_KEY,
    temperature=0.7
)
structured_llm = model_gemini.with_structured_output(Recipe)
chain = prompt | structured_llm
recipe_object = chain.invoke({"dish":"カレー"})

print(type(recipe_object))
print(recipe_object)
```

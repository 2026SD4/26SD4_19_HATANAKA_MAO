```python
import google.generativeai as genai

def generate_recipe_gemini( dish: str ) -> str:
    """ 入力データを引数で受け取りLLMに渡すプロンプトを生成して実行する関数 """
    prompt = f"以下の料理のレシピを考えてください\n料理名: {dish}"
    print (prompt)

    model = genai.GenerativeModel('gemini-flash-latest')
    response = model.generate_content(prompt)
    return response.text
```

```python
recipe_gemini = generate_recipe_gemini("おにぎり")
print(recipe_gemini)
```

```python
import google.generativeai as genai

def generate_spicy_recipe_gemini(dish: str, amount:str, fav:str) -> str:
  context = f"前提条件:\n分量: {amount}\n味の好み: {fav}"
  prompt = f"前提条件を踏まえて、以下のレシピを考えてください。 \n{context}\n料理名:{dish}"

  model = genai.GenerativeModel('gemini-flash-latest')
  response = model.generate_content(prompt)
  return response.text
```

```python
recipe_gemini = generate_spicy_recipe_gemini("クレープ","一人前","甘口")
print(recipe_gemini)
```

```python
from google.colab import userdata
import os

os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_ENDPOINT"] = "https://api.smith.langchain.com"
os.environ["LANGCHAIN_API_KEY"] = userdata.get("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_PROJECT"] = "agentbook"
```

```python
from langchain_google_genai import GoogleGenerativeAI
model = GoogleGenerativeAI( model="gemini-flash-latest",api_key=GOOGLE_API_KEY, temperature=0.7)
output = model.invoke("自己紹介してください")
print(output)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="夏目漱石の「吾輩は猫である」の冒頭からの800文字を表示してください"),
]

for chunk in model.stream(messages):
  print(chunk, end="", flush=True)
```
